```python
# count plot

# count of each category value for hypertention_statue column
status_count = dataSet_df["Hypertension_Status"].value_counts()

# The (normalize = true) returns proportions instead of counts then * 100 multiple them to get percentage value like
status_percentage = dataSet_df['Hypertension_Status'].value_counts(normalize=True) * 100

# Bar chart
plt.figure(figsize=(8,6))
ax = status_count.plot(kind='bar')
status_count.plot(kind='bar')
plt.title("Distribution of Hypertension_Status")
plt.xlabel('Hypertention Status')
plt.ylabel('Number of patients')
plt.xticks(rotation=0)

#  Ai assist with percentage label
# Add percentage label to Bar chart 
for i, count in enumerate(status_count):
    percentage = status_percentage[status_count.index[i]]
    ax.text(
        i,
        count + 10,
        f'({percentage:.1f}%)',
        ha='center'
    )
plt.show()
```

The graph above title distribution of Hypertension_Status where the y-axis has the numbers of patients and x-axis group of hypertension status which show a percentage bar chart of dataset been strongly dominated by stage 2 hypertension with observation of 73.2%.

```python
#  boxplot
plt.figure(figsize=(12,8))
sns.boxplot(
    data=dataSet_df,
    x='Hypertension_Status',
    y='Systolic_BP'
)

# add black dots to indicate observation
sns.stripplot(
    data=dataSet_df,
    x='Hypertension_Status',
    y='Systolic_BP',
    color='black', 
    alpha=0.5,
    size=3, 
)

plt.title('Systolic Blood Pressure by Hypertension Status')
plt.xlabel('Hypertension Status')
plt.ylabel('Systolic Blood Pressure (mmHg)')

plt.show()
```


The box plot title Systolic Blood Pressure by Hypertension Status where systolic_BP is ploted against hypertension status shows the differences in systolic_BP across the three hypertension status groups. the median and overall distribution of systolic_BP increase from normal to Stage1 and State 2, with Stage 2 showing the highest values also high IQR(height of the box), approximatly 0 outlier for unusal value.

## Conclusion and Future work

 - #### Conclusion
This project analyzed a dataset of 1,249 patients to explore patterns and factors associated with hypertension status. The analysis involved data cleaning and preprocessing, exploratory data analysis, statistical correlation, and visualization of relationships between demographic, lifestyle, and health-related variables.

One of the most noticeable findings was the difference in blood pressure measurements across hypertension categories. patients classified as Stage 2 hypertension had substantially higher average systolic and diastolic blood pressure than patients classified as Stage 1 and Normal. This indicates that systolic and diastolic blood pressure were the variables most clearly associated with hypertension status within the analyzed dataset.

The analysis also showed that the dataset was highly imbalanced, with Stage 2 hypertension representing the majority of observations. In addition, most numerical variables showed weak linear correlations with blood pressure, while Weight and BMI demonstrated a strong positive correlation. Age, gender, smoking, alcohol intake, physical activity, and family history did not show substantial differences across hypertension-status groups in this particular dataset.

Overall, the analysis provides useful descriptive insights into the dataset and highlights blood pressure as the most prominent pattern associated with hypertension status. However, the findings should be interpreted within the context of the dataset and should not be considered evidence of causation or a substitute for clinical assessment.

 - #### Key Finding and Implications
   1. the dataset findings tell how systolic and dasolic blood pressure sparks high across the hypertensino group.Stage 2 participants had the highest average systolic and diastolic BP, followed by Stage 1 and Normal participants.

   2. Age showed relatively similar averages across the hypertension groups, so no substantial age difference was observed in this dataset.

   3. Weight and BMI had a strong positive correlation, which is expected because BMI is derived partly from body weight,Most other numerical variables showed weak linear correlations with blood pressure.

   4. The lifestyle and demographic variables analyzed did not show substantial differences across hypertension-status groups.
