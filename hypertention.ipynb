```python
# import necessary liberary or package Globally
import pandas as pd;
import numpy as np;
import matplotlib.pyplot as plt;


# test and see version of package imported Globally
print('pandas :' + pd.__version__)
print('numpy :' + np.__version__)
```

```python
# path to downloaded dataset
URL = './dataSet/Hypertension_Dataset.csv'

# function that load data into pandas data frame
def loadDataSet(URL):
    hypertension_ds = pd.read_csv(URL)
    return  pd.DataFrame(hypertension_ds)



# output downloaded dataset and trucketed from point 5 - 1243
print("Hypertension_Dataset:")
loadDataSet(URL)
```

```python
# To check number of columns and rows 
dataSet_df = loadDataSet(URL).shape
print("numbers of datapoints by variable: \n", dataSet_df)
```

```python
# To see actual names of columns 
dataSet_df = loadDataSet(URL).columns.tolist()
print("columns present in the DataSet: \n", dataSet_df)
```

```python
# To see actual number of columns and dtypes (Numerical and categorical value)
print(" Columns with there Data type and info about the Data: \n")
dataSet_df = loadDataSet(URL).info()
print(dataSet_df)
```

```python
# To check missing values 
print(" Numbers of missing values: \n")
dataSet_df = loadDataSet(URL)
dataSet_df = dataSet_df.isnull().sum()
print(dataSet_df)

```

```python
# To check Duplicate values
dataSet_df = loadDataSet(URL)
dataSet_df = dataSet_df.duplicated()
print("Duplicate values: \n", dataSet_df )

```

```python
# re-comfirmation and removing duplicate rows
dataSet_df = loadDataSet(URL)
dataSet_df = dataSet_df.drop_duplicates() 
print(len(dataSet_df))
```
