```python
# path to downloaded dataset
URL = './dataSet/Hypertension_Dataset.csv'

# function that load data into pandas data frame
def loadDataSet(URL):
    hypertension_ds = pd.read_csv(URL)
    return  pd.DataFrame(hypertension_ds)



# output downloaded dataset and trucketed from point 5 - 1243
print("Hypertension_Dataset:")
loadDataSet(URL)
```

```python
# DataFrame
dataSet_df = loadDataSet(URL)

# count of each category value for hypertention_statue column
staues_count = dataSet_df["Hypertension_Status"].value_counts()

# Bar chart
plt.figure(figsize=(8,6))
staues_count.plot(kind='bar')
plt.title("Distribution of Hypertension_Status")
plt.xlabel('Hypertention Status')
plt.ylabel('Number of patients')
plt.xticks(rotation=0)
plt.show()
```

```python
# Aggregation Comparison check between Hypertension_Status and Systolic_BP

dataSet_df = loadDataSet(URL)
total_hypertention = (dataSet_df
                    .groupby("Hypertension_Status", as_index=False)
                    ['Systolic_BP'].value_counts()
                    .sort_values('Systolic_BP'))

total_hypertention
```

```python
# graph check and visualization

plt.barh(
    total_hypertention["Hypertension_Status"],
    total_hypertention["Systolic_BP"],
    color="#5E6DEE"
)
plt.xlabel('Systolic Blood Pressure')
plt.grid(axis='x', alpha=.8)
plt.tight_layout()
plt.show()
```
