This project aims to analyze patient demographic, lifestyle, and clinical data to investigate patterns and relationships associated with hypertension.
```python
# Transform and encoding target variable

dataSet_df = loadDataSet(URL)
Hypertension_int = {'Normal':0,'Stage 1':1,'Stage 2':2}
dataSet_df['Hypertension_Status'] = dataSet_df['Hypertension_Status'].replace(Hypertension_int).astype('int64')
dataSet_df.sample(5)
```

Transform and encoding the Hypertension_Statue colunm where Normal becomes 0, Stage 1 is 1 and Stage 2 is 2 it easier now to do maths and get the percentage.

```python
# get the count for each values in the column 

dataSet_df['Hypertension_Status'].value_counts()
```

```python
# Percentage distibution of hypertension status
dataSet_df['Hypertension_Status'].value_counts(normalize=True) * 100
```

So 73.2% of the dataset is Stage 2 clearly seen from percentage distibution also 14.3% of the dataset is Normal and 12.5% for Stage 1

```python
# average of numerical value by the hypertension_Statues of value of Stage 2
dataSet_df.select_dtypes(include=np.number).loc[dataSet_df['Hypertension_Status']== 2].mean()
```

```python
# DataFrame
# dataSet_df = loadDataSet(URL)

# count of each category value for hypertention_statue column
status_count = dataSet_df["Hypertension_Status"].value_counts()
status_percentage = dataSet_df['Hypertension_Status'].value_counts(normalize=True) * 100



# Bar chart
plt.figure(figsize=(8,6))
ax = status_count.plot(kind='bar')
status_count.plot(kind='bar')
plt.title("Distribution of Hypertension_Status")
plt.xlabel('Hypertention Status')
plt.ylabel('Number of patients')
plt.xticks(rotation=0)

# Add percentage labe to Bar chart 
for i, count in enumerate(status_count):
    percentage = status_percentage[status_count.index[i]]

    ax.text(
        i,
        count + 10,
        f'({percentage:.1f}%)',
        ha='center'
    )

plt.show()
```

```python
staus_count = dataSet_df["Age"].value_counts()

# Bar chart
plt.figure(figsize=(16,6))
staus_count.plot(kind='bar')
plt.title("Distribution of Hypertension_Status")
plt.xlabel('Hypertention Status')
plt.ylabel('Number of patients')
plt.xticks(rotation=0)
plt.show()
```

```python
# Aggregation Comparison check between Age and Systolic_BP

dataSet_df = loadDataSet(URL)
total_hypertention = (dataSet_df
                    .groupby("Age", as_index=False)
                    ['Systolic_BP'].value_counts()
                    .sort_values('Systolic_BP'))

total_hypertention
# print(total_hypertention.describe())
```

```python
# graph check and visualization

plt.bar(
    total_hypertention["Age"],
    total_hypertention["Systolic_BP"],
    color="#5E6DEE"
)
plt.xlabel('Age')
plt.ylabel('Systolic Blood Pressure')
plt.tight_layout()
plt.show()
```
